**Import Libarary & Dataset**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.svm import SVC
```

**Create a synthetic dataset**

```python
X, y = make_classification(n_samples=100,
                           n_features=2, n_classes=2,
                           n_clusters_per_class=1,
                           n_redundant=0, random_state=42)
```

**Train a Support Vector Machine**

```python
svm_classifier = SVC(kernel='linear', C=1)
svm_classifier.fit(X, y)
```

**Create a mesh grid for the decision boundary**

```python
x_min, x_max= X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02), np.arange(y_min, y_max, 0.02))
Z = svm_classifier.predict(np.c_[xx.ravel(), yy.ravel()])
Z= Z.reshape(xx.shape)
```

**Plot the decision boundary and support vectors**

```python
plt.contourf(xx, yy, Z, alpha=0.4)
plt.scatter(X[:, 0], X[:, 1], c=y, marker='o', edgecolors='k', cmap=plt.cm.Paired)
plt.scatter(svm_classifier.support_vectors_[:, 0], svm_classifier.support_vectors_[:, 1], s=100, facecolors='none', edgecolors='k')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('SVM Decision Boundary')
plt.show()
```
