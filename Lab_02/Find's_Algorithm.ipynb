```python
# Read the data from the CSV file
data = pd.read_csv('/content/drive/MyDrive/Fall_2026 ( L3-T1 )/ICE 3144 - ML Lab/Data sets/Income-Dataset.csv')
print(data)
```

```python
# Making an array of all the attributes
d = np.array(data)[:, :-1]
print("The attributes are:", d)

# Separating the target that has positive and negative examples
target = np.array(data)[:, -1]
print("The target is:", target)
```

```python
# Training function to implement Find-S algorithm
def train(c, t):
    # Find the first positive example
    for i, val in enumerate(t):
        if val == "Yes":
            specific_hypothesis = c[i].copy()
            break

    # Compare all training examples
    for i, val in enumerate(c):
        if t[i] == "Yes":
            for x in range(len(specific_hypothesis)):
                if val[x] != specific_hypothesis[x]:
                    specific_hypothesis[x] = '?'

    return specific_hypothesis
```

```python
# Obtaining the final hypothesis
print("The final hypothesis is:", train(d, target))
```
